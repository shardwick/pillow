# Generating Utility and Simulating UB

```python
# Features: TimeOfDay, SessionDuration, ClickRate, PriorUtility_Score
n_samples = 5000
data = {
    'time_of_day': np.random.randint(0, 24, n_samples),
    'session_duration': np.random.exponential(15, n_samples), # Skewed distribution
    'click_rate': np.random.uniform(0, 1, n_samples),
    'prior_utility': np.random.beta(2, 5, n_samples) # Low prior utility baseline
}

# High ClickRate and low Prior - complex & non-linear
conditions = [
    (data['time_of_day'] > 20) & (data['session_duration'] > 25),
    (data['click_rate'] > 0.7) & (data['prior_utility'] < 0.3)
]
choices = [1, 1]
data['high_utility'] = np.select(conditions, choices, default=0)

df = pd.DataFrame(data)
```
